```python
import numpy as np
#A = np.array([
#    [-5, -1, -3, -1],
#    [-2,  0,  8, -4],
#    [-7, -2,  2, -2],
#    [ 2, -4, -4,  4]
#], dtype=float)
#b = np.array([18, -12, 6, -12], dtype=float)
A = np.array([
    [10, 1, 1],
    [2, 10, 1],
    [2, 2, 10]
])
b = np.array([12, 13, 14], dtype=float)
l = len(A)
```

```python
A_prev = A.astype(float).copy()
L = np.eye(l)
print(L)
for k in range(l - 1):
    A_next = A_prev.copy()

    for i in range(k + 1, l):
        mu_i = A_prev[i, k] / A_prev[k, k]
        print(mu_i)
        L[i, k] = mu_i
        for j in range(k, l):
            A_next[i, j] = (
                A_prev[i, j] - mu_i * A_prev[k, j]
            )
        A_next[i, k] = 0.0
    #print(f"Шаг {k + 1}:")
    #print(A_next)

    A_prev = A_next

print("Итоговая матрица:")
print(A_prev)
print(L)
U = A_prev.copy()
```

```python
np.allclose(A, L @ U)
```

```python
z = np.array([0]*l, dtype=float)
z[0] = b[0]
for i in range(1, l):
    s = 0
    for j in range(i):
        s += L[i, j] * z[j]
    print(s)
    z[i] = b[i] - s
print(z)
```

```python
x = np.array([0]*l, dtype=float)
x[l-1] = z[l-1]/U[l-1][l-1]
for i in range(l-1, -1, -1):
    s = 0
    for j in range(i+1, l):
        s += U[i, j] * x[j]
    x[i] = (z[i] - s) / U[i][i]
print(x)
```
