```python
import numpy as np
A = np.array([
    [-5, -1, -3, -1],
    [-2,  0,  8, -4],
    [-7, -2,  2, -2],
    [ 2, -4, -4,  4]
], dtype=float)
l = len(A)
```

```python
for k in range(1, l):
    A_n = np.array([[0] * l for _ in range(l)], dtype=float)
    for i in range(k + 1, l):
        for j in range(k, l):
            A_n[i][j] = A[i][j] - (A[i][k] * A[k][j]) / A[k][k]
A_n
```
